# solution 1: Using hashmap

```python
class Solution(object):
    def threeSum(self, nums):
        nums.sort()
        result = set()
        map = set()
        n = len(nums)

        for i in range(n-2):
            target = nums[i]
            j = i + 1
            map = set()
            while j < n:
                complement = -target - nums[j]
                if complement in map:
                    result.add((target, nums[j], complement))
                else:
                    map.add(nums[j])
                j+=1
        return [list(x) for x in result]


sol = Solution()
nums = [0, -1, 2, -1, -2]
res = sol.threeSum(nums)
print(res)
```

Time Complexity : $O(N)$

Space Complexity : $O(N)$

## Solution 2: Using two pointer

```python
class Solution(object):
    def threeSum(self, nums):
        nums.sort()
        result = set()
        n = len(nums)
        for i in range(n-2):
            target = nums[i]

            j = i+1
            k = n-1

            while j < k:
                val = target + nums[j] + nums[k]

                if val == 0:
                    result.add((target, nums[j], nums[k]))
                    j += 1
                    k -= 1
                elif val < 0:
                    j += 1
                elif val > 0:
                    k -= 1

        return [list(x) for x in result]

sol = Solution()
nums = [0, -1, 2, -1, -2]
res = sol.threeSum(nums)
print(res)
```

Time Complexity : $O(N)$

Space Complexity : $O(N)$ (However, the actual space complexity of Solution 2 is lower than that of Solution 1)
