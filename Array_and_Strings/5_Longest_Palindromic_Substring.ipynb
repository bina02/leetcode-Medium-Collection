## Solution 1: devide even palindrome and odd palindrome

```python
class Solution(object):
    def longestPalindrome(self, s):  
        if len(s) == 1:
            return s
        longest_palindrome = ''
        lp_left = s[0]
        lp_right = s[0]

        # even palindrome
        for i in range(len(s)-1):
            left = i
            right = i+1
            while left >= 0 and right < len(s) and s[left] == s[right]:
                lp_left = s[left:right+1]
                left -= 1
                right += 1
            longest_palindrome = max(longest_palindrome, lp_left, key=len)

        # odd palindrome
        for i in range(len(s)-2):
            left = i
            right = i+2
            while  left >=0 and right < len(s) and s[left] == s[right]:
                lp_right = s[left:right+1]
                left -= 1
                right += 1
            longest_palindrome = max(longest_palindrome, lp_right, key=len)

        return longest_palindrome


sol = Solution()
s = 'babad'
res = sol.longestPalindrome(s)
print(res)
```

Time Complexity: $O(N^2)$

Space Complexity: $O(1)$

## Solution 1-2: optimized Solution1

```python
class Solution(object):
    def expand(self, left, right, s):
        while left >= 0 and right < len(s) and s[left] == s[right]:
            left -= 1
            right += 1
        return s[left+1:right]

    def longestPalindrome(self, s):  
        if not s:
            return ''
        longest = ''
        for i in range(len(s)):
            even = self.expand(i, i+1, s)
            odd = self.expand(i, i, s)
            longest = max(longest, even, odd, key=len)
        return longest


sol = Solution()
s = 'a'
res = sol.longestPalindrome(s)
print(res)
```

Time Complexity: $O(N^2)$

Space Complexity: $O(1)$

## Solution 2: Reusing palindrome

```python
class Solution(object):
    def longestPalindrome(self, s):
        n = len(s)
        matrix = [[False]*n for _ in range(n)]
        if n < 2:
            return s

        for i in range(n):
            matrix[i][i] = True
            start = i
            max_len = 1

        for i in range(n-1):
            if s[i] == s[i+1]:
                matrix[i][i+1] = True
                start = i
                max_len = 2

        for length in range(3, n+1):
            for i in range(n-length+1):
                j = i+length-1
                if s[i] == s[j] and matrix[i+1][j-1]:
                    matrix[i][j] = True
                    start = i
                    max_len = length

        return s[start:start+max_len]

sol = Solution()
s = "cbbd"
res = sol.longestPalindrome(s)
print(res)
```

Time Complexity: $O(N^2)$

Space Complexity: $O(N^2)$
