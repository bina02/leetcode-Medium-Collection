## Solution 1: Saving Zero index

```python
class Solution(object):
    def setZeroes(self, matrix):
        zero_index = []
        for r in range(len(matrix)):
            for c in range(len(matrix[r])):
                if matrix[r][c] == 0:
                    zero_index.append([r,c])

        for z_i in zero_index:
            row = z_i[0]
            col = z_i[1]

            matrix[row] = [0] * len(matrix[0])
            for r in range(len(matrix)):
                matrix[r][col] = 0
        
        return matrix

sol = Solution()
matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
res = sol.setZeroes(matrix)
print(res)
```

Time Complexity: $O(M \times N)$

Space Complexity: $O(M \times N)$

## Solution 2: Using marker

```python
class Solution(object):
    def setZeroes(self, matrix):   
        row = len(matrix)
        col = len(matrix[0])

        first_row_zero = False
        first_col_zero = False
        # Checking zero in first row&col
        for c in range(col):
            if matrix[0][c] == 0:
                first_row_zero = True
        for r in range(row):
           if matrix[r][0] == 0:
               first_col_zero = True

        # marking
        for r in range(1, row):
            for c in range(1, col):
                if matrix[r][c] == 0:
                    matrix[r][0] = 0
                    matrix[0][c] = 0

        # convert rows&cols to zero (except first row&col)
        for r in range(1, row):
            if matrix[r][0] == 0:
                matrix[r] = [0]*col
        for c in range(1, col):
            if matrix[0][c] == 0:
                for r in range(row):
                    matrix[r][c] = 0

        # convert first row&col to 0
        if first_col_zero:
            for r in range(row):
                matrix[r][0] = 0
        if first_row_zero:
            matrix[0] = [0]*col
        return matrix    


sol = Solution()
matrix = [[0,1,2,0],[3,4,5,2],[1,3,1,5]]
res = sol.setZeroes(matrix)
print(res)
```

Time Complexity: $O(M \times N)$

Space Complexity: $O(1)$
