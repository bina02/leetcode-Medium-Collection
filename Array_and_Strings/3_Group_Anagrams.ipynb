## Solution 1: Brute Force & Sorting

```python
class Solution(object):
    def checkAnagrams(self, str1, str2):
        str1 = list(str1)
        str2 = list(str2)
        str1.sort()
        str2.sort()
        return str1 == str2

    def groupAnagrams(self, strs):
        anagram_list = [[strs[0]]]

        for i in range(1, len(strs)):
            for j in range(len(anagram_list)):
                if self.checkAnagrams(strs[i], anagram_list[j][0]):
                    anagram_list[j].append(strs[i])
                    break
            else:
                anagram_list.append([strs[i]])

        return anagram_list

sol = Solution()
strs = ["eat","tea","tan","ate","nat","bat"]
res = sol.groupAnagrams(strs)
print(res)
```

Time Complexity: $O(N^2 \times K\text{log}K)$ ($K$ is length of word)

Space Complexity: $O(N)$

## Solution 2: Using hashmap

```python
class Solution(object):
    def groupAnagrams(self, strs):
        map = {}    
        for s in strs:
            sorted_str = ''.join(sorted(s))
            if sorted_str not in map:
                map[sorted_str] = []
            map[sorted_str].append(s)
        return list(map.values())

sol = Solution()
strs = ["eat","tea","tan","ate","nat","bat"]
res = sol.groupAnagrams(strs)
print(res)
```

Time Complexity: $O(N \times K\text{log}K)$ ($K$ is length of word)

Space Complexity: $O(N)$
