## Solution 1: Two pointer

```python
class Solution(object):
    def RLE(self, n):
        res = []
        left = 0
        while left < len(n):
            right = left
            while right < len(n) and n[left] == n[right]:
                right += 1

            count = str(right - left)
            s = n[left]
            res.append(count)
            res.append(s)
            left = right

        return ''.join(res)
    
    def countAndSay(self, n):
        n = int(n)
        res = '1'
        for _ in range(n-1):
            res = self.RLE(res)

        return res

sol = Solution()
n = '4'
res = sol.countAndSay(n)
print(res)
```

Time Complexity: $O(\lambda^n)$ ($\lambda$ = 1.303...)

Space Complexity: $O(\lambda^n)$ ($\lambda$ = 1.303...)

## Solution 2: Using recursion

```python
class Solution(object):
    def RLE(self, n):
        res = []
        left = 0
        while left < len(n):
            right = left
            while right < len(n) and n[left] == n[right]:
                right += 1

            count = str(right - left)
            s = n[left]
            res.append(count)
            res.append(s)
            left = right

        return ''.join(res)

    def countAndSay(self, n):
        if n == 1:
            return '1'

        return self.RLE(self.countAndSay(n-1))

        
sol = Solution()
n = 4
res = sol.countAndSay(n)
print(res)
```

Time Complexity: $O(\lambda^n)$ ($\lambda$ = 1.303...)

Space Complexity: $O(\lambda^n)$ ($\lambda$ = 1.303...) +($O(N)$ stack)
