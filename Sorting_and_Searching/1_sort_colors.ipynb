## Solution 1: Brute force

```python
class Solution(object):
    def sortColors(self, nums):
        for i in range(len(nums)):
            for j in range(i+1, len(nums)):
                if nums[i] > nums[j]:
                    nums[i], nums[j] = nums[j], nums[i]
        return nums
sol = Solution()
nums = [2,2,0,0,1,1]
res = sol.sortColors(nums)
print(res)
        
```

Time Complexity: $O(N^2)$

Space Complexity: $O(1)$

## Solution 2: Saving count of numbers

```python
class Solution(object):
    def sortColors(self, nums):
        zero_count = 0
        one_count = 0
        two_count = 0
        for i in range(len(nums)):
            if nums[i] == 0:
                zero_count += 1
            if nums[i] == 1:
                one_count += 1
            if nums[i] == 2:
                two_count += 1

        for i in range(zero_count):
            nums[i] = 0
        for i in range(zero_count, zero_count+one_count):
            nums[i] = 1
        for i in range(zero_count+one_count, zero_count+one_count+two_count):
            nums[i] = 2

        return nums


            
sol = Solution()
nums = [2,2,1,0,0,1,1]
res = sol.sortColors(nums)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$

## Soution 3: Using three pointer (One pass)

```python
class Solution(object):
    def sortColors(self, nums):
        low, mid, high = 0, 0, len(nums)-1
        while mid <= high:
            if nums[mid] == 0:
                nums[low], nums[mid] = nums[mid], nums[low]
                low += 1
                mid += 1
            elif nums[mid] == 1:
                mid += 1
            elif nums[mid] == 2:
                nums[mid], nums[high] = nums[high], nums[mid]
                high -= 1
        return nums
    
sol = Solution()
nums = [2,2,1,0,0,1,1]
res = sol.sortColors(nums)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$
