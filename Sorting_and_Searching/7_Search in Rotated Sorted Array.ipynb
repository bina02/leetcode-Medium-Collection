## Solution 1: Binary search with find max_idx

```python
class Solution(object):
    def search(self, nums, target):
        if not nums:
            return -1
        if len(nums) == 1:
            if nums[0] == target:
                return 0
            else:
                return -1
            
        def find_max_idx(li):
            left = 0
            right = len(nums) -1
            if li[left] <= li[right]:
                return right
            
            while left < right:
                mid = (left + right) // 2

                if nums[mid] > nums[mid+1]:
                    return mid
                if nums[mid] < nums[mid-1]:
                    return mid-1

                if nums[mid] >= nums[left]:
                    left = mid + 1
                elif nums[mid] < nums[left]:
                    right = mid - 1
         
        max_idx = find_max_idx(nums)

        if max_idx == len(nums)-1:
            left = 0
            right = len(nums)-1
        
        elif target <= nums[-1]:
            left = max_idx+1
            right = len(nums)-1
        
        else:
            left = 0
            right = max_idx

        
        while left <= right:
            mid = (left + right) // 2
            if target > nums[mid]:
                left = mid + 1
            elif target < nums[mid]:
                right = mid - 1
            else:
                return mid
        return -1

sol = Solution()
nums = [5,6,7,8,1,2,3,4]
target = 1
res = sol.search(nums, target)
print(res)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(1)$

## Solution 2: One half is always sorted

```python
class Solution(object):
    def search(self, nums, target):
        left = 0
        right = len(nums) - 1

        while left <= right:
            mid = (left + right) // 2

            if nums[mid] == target:
                return mid
            
            if nums[left] <= nums[mid]:
                if nums[left] <= target < nums[mid]:
                    right = mid - 1
                else:
                    left = mid + 1
            
            else:
                if nums[mid] < target <= nums[right]:
                    left = mid + 1
                else:
                    right = mid - 1
            
        return -1

sol = Solution()
nums = [5,6,7,8,1,2,3,4]
target = 1
res = sol.search(nums, target)
print(res)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(1)$
