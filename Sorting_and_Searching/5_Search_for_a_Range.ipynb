## Solution 1: Brute force

```python
class Solution(object):
    def searchRange(self, nums, target):
        idx_li = []
        for i in range(len(nums)):
            if nums[i] == target:
                idx_li.append(i)
                end = idx_li[0]

                while end+1<len(nums) and nums[end+1] == target:
                    end += 1
                idx_li.append(end)
                return idx_li
        
        return [-1, -1]

sol = Solution()
nums = [5,7,7,8,8,8,10]
res = sol.searchRange(nums, 10)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$

## Solution 2: Binary search

```python
class Solution(object):
    def searchRange(self, nums, target):
        def find_start_end(is_first):
            left = 0
            right = len(nums)-1
            bound = -1

            while left <= right:
                mid = (left + right) // 2
                if nums[mid] < target:
                    left = mid+1
                elif nums[mid] > target:
                    right = mid-1
                else:
                    bound = mid
                    if is_first:
                        right = mid-1
                    else:
                        left = mid+1
            return bound

        return [find_start_end(True), find_start_end(False)]



sol = Solution()
nums = [5,7,7,8,8,8,10,14]
res = sol.searchRange(nums, 13)
print(res)
```

Time Complexity: mean: $O(\text{log}N)$, worst: $O(N)$

Space Complexity: $O(1)$
