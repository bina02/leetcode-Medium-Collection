## Solution 1: Binary serach per row

```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        if not matrix[0]:
            return False

        def binarySearch(nums, target):
            left = 0
            right = len(nums) - 1

            while left <= right:
                mid = (left + right) // 2
                if nums[mid] == target:
                    return True
                elif nums[mid] < target:
                    left = mid + 1
                else:
                    right = mid - 1

        for row in matrix:
            if row[0] <= target <= row[-1]:
                if binarySearch(row, target):
                    return True

        return False

sol = Solution()
matrix = [[]]
target = 1
res = sol.searchMatrix(matrix, target)
print(res)
```

Time Complexity: $O(m\text{log}N)$

Space Complexity: $O(1)$

## Solution 2: Top-right / Bottom-Left search

```python
class Solution(object):
    def searchMatrix(self, matrix, target):
        if not matrix[0]:
            return False

        row = 0
        col = len(matrix[0]) - 1

        while row < len(matrix) and col >= 0:
            if matrix[row][col] == target:
                return True
            elif matrix[row][col] < target:
                row += 1
            else:
                col -= 1
            
        return False

sol = Solution()
matrix = [[1,4,7,11,15],[2,5,8,12,19],[3,6,9,16,22],[10,13,14,17,24],[18,21,23,26,30]]
target = 31
res = sol.searchMatrix(matrix, target)
print(res)
```

Time Complexity: $O(M+N)$

Space Complexity: $O(1)$
