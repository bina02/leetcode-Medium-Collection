## Solution 1: Using most_common(k)

```python
from collections import Counter

class Solution(object):
    def topKFrequent(self, nums, k):
        res_li = []
        counter = Counter(nums)
        most_common_li = counter.most_common(k)
        for key, val in most_common_li:
            res_li.append(key)
        return res_li


sol = Solution()
nums = [1,2,1,2,1,2,3,1,3,2]
res = sol.topKFrequent(nums, 2)
print(res)
```

Time Complexity: $O(N \text{log}k)$

Space Complexity: $O(N)$

## Solution 2: Using bucket sort

```python
from collections import Counter

class Solution(object):
    def topKFrequent(self, nums, k):
        res_li = []
        counter = Counter(nums)
        buckets = [[] for _ in range(len(nums)+1)]

        for num, freq in counter.items():
            buckets[freq].append(num)

        for i in range(len(buckets)-1, 0, -1):
            if not buckets[i]:
                continue
            for num in buckets[i]:
                res_li.append(num)
                if len(res_li) == k:
                    return res_li

sol = Solution()
nums = [1,1,2,2,2,3,3,3,3]
res = sol.topKFrequent(nums, 2)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$
