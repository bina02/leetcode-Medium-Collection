## Solution 1: Using sorting

```python
class Solution(object):
    def findKthLargest(self, nums, k):
        sorted_nums = sorted(nums)
        return sorted_nums[-k]

sol = Solution()
nums = [3,2,1,5,6,4]
k = 2
res = sol.findKthLargest(nums, k)
print(res)
        
```

Time Complexity: $O(N\text{log}N)$

Space complexity: $O(N)$

## Solution 2: Using heap

```python
import heapq

class Solution(object):
    def findKthLargest(self, nums, k):
        heap = nums[:k]
        heapq.heapify(heap)

        for num in nums[k:]:
            if heap[0] < num:
                heapq.heappushpop(heap, num)

        return heap[0]

sol = Solution()
nums = [3,2,3,1,2,4,5,5,6]
k = 4
res = sol.findKthLargest(nums, k)
print(res)
        
```

Time Complexity: $O(N\times{log}k)$

Time Complexity: $O(k)$

## Solution 3: Quick select

```python
import random

class Solution(object):
    def findKthLargest(self, nums, k):
        pivot = random.choice(nums)

        left = [x for x in nums if x < pivot]
        mid = [x for x in nums if x == pivot]
        right = [x for x in nums if x > pivot]

        if len(right) >= k:
            return self.findKthLargest(right, k)
        elif len(right)+len(mid) < k:
            return self.findKthLargest(left, k-len(right)-len(mid))
        else:
            return pivot
    
sol = Solution()
nums = [3,2,3,1,2,4,5,5,6]
k = 2
res = sol.findKthLargest(nums, k)
print(res)
```

Time Complexity: $\text{mean: }O(N)$ $\text{worst: }O(N^2)$

Time Complexity: $O(N)$
