## Solution 1: Using itertools library

```python
from itertools import product
class Solution(object):
    def letterCombinations(self, digits):
        if digits == '':
            return []
        res_li = []
        digits_dict = {
            '2': 'abc',
            '3': 'def',
            '4': 'ghi',
            '5': 'jkl',
            '6': 'mno',
            '7': 'pqrs',
            '8': 'tuv',
            '9': 'wxyz'
        }

        char_li = list(digits)
        digits_li = [digits_dict[char] for char in char_li]
        res_li = [''.join(li) for li in product(*digits_li)]
        return res_li

sol = Solution()
digits = '234'
res = sol.letterCombinations(digits)
print(res)
```

Time Complexity: $O(4^N \times N)$

Space Complexity: $O(4^N \times N)$

## Solution 2: DFS

```python
class Solution(object):
    def letterCombinations(self, digits):
        if digits == '':
            return []
        res_li = []
        digits_dict = {
            '2': 'abc',
            '3': 'def',
            '4': 'ghi',
            '5': 'jkl',
            '6': 'mno',
            '7': 'pqrs',
            '8': 'tuv',
            '9': 'wxyz'
        }

        def dfs(index, path):
            if len(path) == len(digits):
                res_li.append(path)
                return 
            for char in digits_dict[digits[index]]:
                dfs(index+1 , path + char)

        dfs(0, '')
        return res_li
            
        
sol = Solution()
digits = '234'
res = sol.letterCombinations(digits)
print(res)
```

Time Complexity: $O(4^N \times N)$

Space Complexity: $O(4^N \times N)$

## Solution 3: BFS

```python
from collections import deque
class Solution(object):
    def letterCombinations(self, digits):
        if digits == '':
            return []
        res_li = []
        digits_dict = {
            '2': ['a', 'b', 'c'],
            '3': ['d', 'e', 'f'],
            '4': ['g', 'h', 'i'],
            '5': ['j', 'k', 'l'],
            '6': ['m', 'n', 'o'],
            '7': ['p', 'q', 'r', 's'],
            '8': ['t', 'u', 'v'],
            '9': ['w', 'x', 'y', 'z']
        }

        queue = deque(digits_dict[digits[0]])
        index = 0
        while queue:
            if index+1 == len(digits):
                return list(queue)
            
            for _ in range(len(queue)):
                current_str = queue.popleft()
                for char in digits_dict[digits[index+1]]:
                    queue.append(current_str+char)
            index += 1

sol = Solution()
digits = '23'
res = sol.letterCombinations(digits)
print(res)
```

Time Complexity: $O(4^N \times N)$

Space Complexity: $O(4^N \times N)$
