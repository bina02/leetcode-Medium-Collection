## Solution 1: Backtracking

```python
class Solution(object):
    def exist(self, board, word):
        row = len(board)
        col = len(board[0])

        def dfs(r, c, idx):
            if idx == len(word):
                return True
            
            if r >= row or r < 0 or c >= col or c < 0 or idx >= len(word) or board[r][c] != word[idx]:
                return False
            
            temp = board[r][c]
            board[r][c] = '#'
            found = dfs(r+1, c, idx+1) or dfs(r-1, c, idx+1) or dfs(r, c+1, idx+1) or dfs(r, c-1, idx+1)
            return found

        for r in range(row):
            for c in range(col):
                if dfs(r, c, 0):
                    return True
        return False


            
sol = Solution()
board = [["A","B","C","E"],["S","F","C","S"],["A","D","E","E"]]
word = 'SEE'
res = sol.exist(board, word)
print(res)

        
```

Time Complexity: $O(N \times M \times3^L)$

Space Complexity: $O(L)$ ($L$ is depth of call stack)
