## Solution 1: Backtracking

```python
class Solution(object):
    def permute(self, nums):
        res_li = []
        def dfs(path):
            if len(path) == len(nums):
                res_li.append(path)
                return

            for num in nums:
                if num not in path:
                    dfs(path + [num])

        dfs([])
        return res_li


sol = Solution()
nums = [1,2,3]
res = sol.permute(nums)
```

Time Complexity: $O(N \times N!)$

Space Complexity: $O(N!)$

## Solution 2: Backtracking Using hashmap

```python
class Solution(object):
    def permute(self, nums):
        res_li = []
        visited = set()
        def dfs(path):
            if len(path) == len(nums):
                res_li.append(path[:])
                return

            for num in nums:
                if num not in visited:
                    visited.add(num)
                    path.append(num)
                    dfs(path)
                    path.pop()
                    visited.remove(num)
                    
        dfs([])
        return res_li

sol = Solution()
nums = [1,2,3]
res = sol.permute(nums)
print(res)
```

Time Complexity: $O(N \times N!)$

Space Complexity: $O(N!)$
