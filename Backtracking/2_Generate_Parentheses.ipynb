## Solutio 1: Brute force

```python
class Solution(object):
    def isValid(self, s):
        stack = []
        for par in s:
            if par == '(':
                stack.append('(')
            if par == ')':
                if stack:
                    stack.pop()
                else:
                    return False
        if stack:
            return False
        else:
            return True

    def generateParenthesis(self, n):
        res_li = []
        def dfs(path):
            if len(path) == 2*n:
                if self.isValid(path):
                    res_li.append(path)
                return

            for par in ['(', ')']:
                dfs(path + par)

        dfs('(')
        return res_li

sol = Solution()
res = sol.generateParenthesis(4)
print(res)
```

Time Complexity: $O(4^N \times N)$

Space Complexity: $O(4^N \times N)$

```python
class Solution(object):
    def generateParenthesis(self, n):
        res_li = []
        def dfs(path, open_count, close_count):
            if len(path) == 2*n:
                res_li.append(path)
                return
            if open_count < n:
                dfs(path + '(', open_count+1, close_count)
            if close_count < open_count:
                dfs(path + ')', open_count, close_count+ 1)

        dfs('(', 1, 0)
        return res_li

sol = Solution()
res = sol.generateParenthesis(2)
print(res)
```

Time Complexity: $O(C_n)$

Space Complexity: $O(C_n)$
