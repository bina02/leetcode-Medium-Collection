## Solution 1: BFS (+pooling)

```python
from collections import deque
class Solution(object):
    def coinChange(self, coins, amount):
        if amount == 0:
            return 0

        queue = deque([0])
        visited = {0}
        depth = 0
        
        while queue:
            for _ in range(len(queue)):
                curr_amount = queue.popleft()

                if curr_amount == amount:
                    return depth
                
                for coin in coins:
                    amount_sum = curr_amount + coin
                    if amount_sum <= amount and amount_sum not in visited:
                        visited.add(amount_sum)
                        queue.append(amount_sum)

            depth += 1

        return -1

sol = Solution()
coins, amount = [1,2,5], 13
res = sol.coinChange(coins, amount)
print(res)
```

Time Complexity: $O(C \times A)$ ($C$ is coin, $A$ is amount)

Space Complexity: $O(A)$

## Solution 2: DP

```python
from collections import deque
class Solution(object):
    def coinChange(self, coins, amount):
        dp = [float('inf')] * (amount+1)
        dp[0] = 0

        for i in range(1, amount+1):
            for coin in coins:
                if (i-coin)>= 0:
                    dp[i] = min(dp[i], dp[i-coin] + 1)

        if dp[amount] != float('inf'):
            return dp[amount]
        else:
            return -1
        
sol = Solution()
coins, amount = [2,4,5], 12
res = sol.coinChange(coins, amount)
print(res)
```

Time Complexity: $O(C \times A)$ ($C$ is coin, $A$ is amount)

Space Complexity: $O(A)$
