## Solution 1: Permutation with repetition

```python
import math

class Solution(object):
    def uniquePaths(self, m, n):
        m -= 1
        n -= 1
        
        if m < 1 or n < 1:
            return 0
        if m == 1 or n == 1:
            return 1

        m_factorial = math.factorial(m)
        n_factorial = math.factorial(n)
        total_factorial = math.factorial(m+n)
        return total_factorial // (m_factorial*n_factorial)

sol = Solution()
m, n = 3, 7
res = sol.uniquePaths(m, n)
print(res)
```

Time Complexity: $O(M+N)$

Space Complexity: $O(1)$

## Solution 2: DFS

```python
class Solution(object):
    def uniquePaths(self, m, n):
        if m < 1 or n < 1:
            return 0
        if m == 1 or n == 1:
            return 1

        self.count = 0
        def dfs(x, y):
            if x > m or y > n:
                return 

            dfs(x+1, y)
            dfs(x, y+1)
            
            if x == m and y == n:
                self.count += 1
        dfs(1, 1)
        return self.count

sol = Solution()
m, n = 3, 7
res = sol.uniquePaths(m, n)
print(res)
```

Time Complexity: $O(2^{m+n})$

Space Complexity: $O(M+N)$

## Solution 3: DP

```python
class Solution(object):
    def uniquePaths(self, m, n):
        dp = [1] * n

        for r in range(1, m):
            for c in range(1, n):
                dp[c] += dp[c-1]

        return dp[-1]

sol = Solution()
m, n = 3, 7
res = sol.uniquePaths(m, n)
print(res)
```

Time Complexity: $O(M \times N)$

Space Complexity: $O(N)$
