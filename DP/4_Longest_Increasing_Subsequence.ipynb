## Solution 1: DP

```python
class Solution(object):
    def lengthOfLIS(self, nums):
        if not nums:
            return 0

        dp = [1] * len(nums)

        for i in range(len(nums)):
            for j in range(i):
                if nums[j] < nums[i]:
                    dp[i] = max(dp[i], dp[j]+1)

        return max(dp)

sol = Solution()
nums = [10,9,2,5,3,7,101,18]
res = sol.lengthOfLIS(nums)
print(res)
```

Time Complexity: $O(N^2)$

Space Complexity: $O(N)$

## Solution 2: Binary search

```python
import bisect
class Solution(object):
    def lengthOfLIS(self, nums):
        if not nums:
            return 0
        
        lis = []
        for num in nums:
            idx = bisect.bisect_left(lis, num)
            if idx == len(lis):
                lis.append(num)
            else:
                lis[idx] = num

        return len(lis)

sol = Solution()
nums = [10,11,2,5,3,7,101,18]
res = sol.lengthOfLIS(nums)
print(res)
```

Time Complexity: $O(N \text{log}N)$

Space Complexity: $O(N)$
