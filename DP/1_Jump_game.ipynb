## Solution 1: DFS

```python
class Solution(object):
    def canJump(self, nums):
        if not nums:
            return False

        def dfs(index, nums):
            if index >= len(nums)-1:
                return True
            
            for i in range(1, nums[index]+1):
                if dfs(index+i, nums):
                    return True
            
            return False
        
        return dfs(0, nums)

sol = Solution()
nums = [1,2,1,0,3]
res = sol.canJump(nums)
print(res)
```

Time Complexity: $O(M^N)$

Space Complexity: $O(N)$

## Solution 2: greedy search

```python
class Solution(object):
    def canJump(self, nums):
        if not nums:
            return False
        
        max_reach = 0
        for i, jump in enumerate(nums):
            if i > max_reach:
                return False

            max_reach = max(max_reach, i+nums[i])

            if max_reach >= len(nums)-1:
                return True
        return True

sol = Solution()
nums = [1,2,1,3,3]
res = sol.canJump(nums)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$
