## Preset

```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

node_c3 = ListNode(5)
node_c2 = ListNode(4, node_c3)
node_common = ListNode(8, node_c2)

node_a2 = ListNode(1, node_common)
headA = ListNode(4, node_a2)

node_b3 = ListNode(1, node_common)
node_b2 = ListNode(6, node_b3)
headB = ListNode(5, node_b2)

def displayList(head):
        values = []
        current_node = head
        while current_node is not None:
            values.append(str(current_node.val))
            current_node = current_node.next
        print(" -> ".join(values))
```

## Solution1. Using Hashmap

```python
class Solution(object):
    def getIntersectionNode(self, headA, headB):
        map = set()
        A_node = headA
        B_node = headB

        while A_node:
            map.add(A_node)
            A_node = A_node.next

        while B_node:
            if B_node in map:
                return B_node
            B_node = B_node.next

        return None

sol = Solution()
res = sol.getIntersectionNode(headA, headB)
print('listA')
displayList(headA)
print('listB')
displayList(headB)
print('common')
displayList(res)
```

Time complexity: $O(M+N)$ (listA: $M$, listB: $N$)

Space Complexity: $O(M)$


## Solution2. Connecting Two list

```python
class Solution(object):
    def getIntersectionNode(self, headA, headB):
        A_node = headA
        B_node = headB

        while A_node != B_node:
            A_node = A_node.next if A_node else headB
            B_node = B_node.next if B_node else headA

        return A_node

sol = Solution()
res = sol.getIntersectionNode(headA, headB)
print('listA')
displayList(headA)
print('listB')
displayList(headB)
print('common')
displayList(res)
```

Time complexity: $O(M+N)$ (listA: $M$, listB: $N$)

Space Complexity: $O(1)$
