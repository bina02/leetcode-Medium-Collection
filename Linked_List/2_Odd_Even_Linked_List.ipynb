## Preset

```python
class ListNode():
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

node6 = ListNode(6)
node5 = ListNode(5, node6)
node4 = ListNode(4, node5)
node3 = ListNode(3, node4)
node2 = ListNode(2, node3)
node1 = ListNode(1, node2)

def displayList(head):
        values = []
        current_node = head
        while current_node is not None:
            values.append(str(current_node.val))
            current_node = current_node.next
        print(" -> ".join(values))
```

## Solution 1: connecting evenList and oddList

```python
import copy
class Solution(object):
    def oddEvenList(self, head):
        if not head or not head.next:
             return head
        
        head_cp = copy.deepcopy(head)
        even_head = head.next
        current_node = even_head
        while current_node.next and current_node.next.next:
            current_node.next = current_node.next.next
            current_node = current_node.next
        if current_node.next and not current_node.next.next:
            current_node.next = None

        odd_head = head_cp
        current_node = odd_head
        while current_node.next and current_node.next.next:
            current_node.next = current_node.next.next
            current_node = current_node.next

        current_node.next = even_head

        return odd_head
sol = Solution()
print('before')
displayList(node1)
res = sol.oddEvenList(node1)
print('after')
displayList(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$

```python
class Solution(object):
    def oddEvenList(self, head):
        if not head or not head.next:
            return head

        even_head = head.next
        even = head.next
        odd = head

        while even and even.next:
            odd.next = even.next
            odd = odd.next
            even.next = odd.next
            even = even.next

        odd.next = even_head
        return head
    
sol = Solution()
print('before')
displayList(node1)
res = sol.oddEvenList(node1)
print('after')
displayList(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$
