## Preset

```python
class ListNode(object):
    def __init__(self, val=0, next=None):
        self.val = val
        self.next = next

node13 = ListNode(val=0, next=None)
node12 = ListNode(val=4, next=node13)
node11 = ListNode(val=2, next=node12)

node24 = ListNode(val=9, next=None)
node23 = ListNode(val=4, next=node24)
node22 = ListNode(val=6, next=node23)
node21 = ListNode(val=5, next=node22)
```

## Solution 1: linkedlist -> list -> int -> list -> linkedlist

```python
class Solution(object):
    def convert_to_int(self, head):
        res = []
        current_node = head
        while current_node != None:
            res.append(current_node.val)
            current_node = current_node.next
        res = res[::-1]
        return int(''.join(map(str, res)))

    def addTwoNumbers(self, l1, l2):
        sum_head = self.convert_to_int(l1) + self.convert_to_int(l2)
        sum_head = [int(digit) for digit in str(sum_head)]
        dummy = ListNode(0)
        current_node = dummy
        for i in range(len(sum_head)-1, -1, -1):
            current_node.next = ListNode(val=sum_head[i])
            current_node = current_node.next
        current_node = dummy.next

        return dummy.next

sol = Solution()
res = sol.addTwoNumbers(node11, node21)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$


## Solution 2: Two-Pointer Linked List Traversal

```python
class Solution(object):
    def addTwoNumbers(self, l1, l2):
        dummy = ListNode(0)
        current_node = dummy
        carry = 0
        while l1 or l2 or carry:
            val1 = l1.val if l1 else 0
            val2 = l2.val if l2 else 0

            total = val1 + val2 + carry
            carry = total // 10

            current_node.next = ListNode(total % 10)
            current_node = current_node.next
            if l1: l1 = l1.next
            if l2: l2 = l2.next
        return dummy.next

sol = Solution()
res = sol.addTwoNumbers(node11, node21)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$
