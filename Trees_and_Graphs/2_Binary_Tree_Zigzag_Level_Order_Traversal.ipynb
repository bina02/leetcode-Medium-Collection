## Preset

```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(3)
node4 = TreeNode(4)
node5 = TreeNode(5)
node6 = TreeNode(6)
node7 = TreeNode(7)
node8 = TreeNode(8)
node9 = TreeNode(9)

node1.left = node2
node1.right = node3

node2.left = node4
node2.right = node5

node5.left = node6
node5.right = node7

node3.right = node8

node8.left = node9

root = node1
```

## Solution 1: Using stack

```python
class Solution(object):
    def zigzagLevelOrder(self, root):
        if root is None:
            return []
        
        res_li = []
        current_level = [root]
        next_level = []
        current_node = root
        switch = True

        while current_level: 
            sub_li = []
            while current_level:
                current_node = current_level.pop()
                if current_node is None:
                    continue

                if switch:
                    sub_li.append(current_node.val)
                    next_level.append(current_node.left)
                    next_level.append(current_node.right)

                else:
                    sub_li.append(current_node.val)
                    next_level.append(current_node.right)
                    next_level.append(current_node.left)                    
            if sub_li:
                res_li.append(sub_li)
            switch = not switch

            current_level = next_level
            next_level = []

        return res_li

sol = Solution()
res = sol.zigzagLevelOrder(root)
print(res)

```

Time Complexity: $O(N)$


Space Complexity: $O(L)$ (where L is the maximum width of the tree)

## Solution 2: Using queue

```python
from collections import deque

class Solution(object):
    def zigzagLevelOrder(self, root):
        if root is None:
            return []
        
        queue = deque([root])
        res_li = []
        switch = True

        while queue:
            level_size = len(queue)
            sub_li = deque()

            for _ in range(level_size):
                current_node = queue.pop()
                if switch:
                    sub_li.appendleft(current_node.val)
                else:
                    sub_li.append(current_node.val)

                if current_node.right:
                    queue.appendleft(current_node.right)
                if current_node.left:
                    queue.appendleft(current_node.left)

            res_li.append(list(sub_li))
            switch = not switch

        return res_li
            

sol = Solution()
res = sol.zigzagLevelOrder(root)
print(res)

```

Time Complexity: $O(N)$


Space Complexity: $O(L)$ (where L is the maximum width of the tree)
