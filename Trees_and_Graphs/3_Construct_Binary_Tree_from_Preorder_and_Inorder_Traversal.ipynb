## Preset

```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

## Solution 1: divide list (slicing)

```python
class Solution(object):
    def buildTree(self, preorder, inorder):
        if not preorder and not inorder:
            return None

        root = TreeNode(val=preorder[0])

        devide_idx = inorder.index(root.val)
        left_inorder = inorder[:devide_idx]
        right_inorder = inorder[1+devide_idx:]

        left_preorder = preorder[1:1+len(left_inorder)]
        right_preorder = preorder[1+len(left_inorder):]

        root.left = self.buildTree(left_preorder, left_inorder)
        root.right = self.buildTree(right_preorder, right_inorder)
        return root

sol = Solution()
preorder = [3,9,20,15,7]
inorder = [9,3,15,20,7]
res = sol.buildTree(preorder, inorder)
print(res)
```

Time Complexity: $O(N^2)$

Space Complexity: $O(N)$

## Solution 1: Using hashmap (not using slicing)

```python
class Solution(object):
    def buildTree(self, preorder, inorder):
        in_dict = {}
        for i in range(len(preorder)):
            in_dict[inorder[i]] = i
        self.pre_idx = 0

        def build(in_start, in_end):
            if in_start > in_end:
                return None
            
            root = TreeNode(val=preorder[self.pre_idx])
            self.pre_idx += 1
            devide_idx = in_dict[root.val]

            root.left = build(in_start, devide_idx-1)
            root.right = build(devide_idx+1, in_end)

            return root

        return build(0, len(preorder)-1)
        
sol = Solution()
preorder = [3,9,20,15,7]
inorder = [9,3,15,20,7]
res = sol.buildTree(preorder, inorder)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$
