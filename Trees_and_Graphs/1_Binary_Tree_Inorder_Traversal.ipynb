## Preset

```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right

node1 = TreeNode(1)
node2 = TreeNode(2)
node3 = TreeNode(3)
node4 = TreeNode(4)
node5 = TreeNode(5)
node6 = TreeNode(6)
node7 = TreeNode(7)
node8 = TreeNode(8)
node9 = TreeNode(9)

node1.left = node2
node1.right = node3

node2.left = node4
node2.right = node5

node5.left = node6
node5.right = node7

node3.right = node8

node8.left = node9

root = node1
```

## Solution 1: DFS

```python
class Solution(object):            
    def inorderTraversal(self, root):
        val_li = []
        def dfs(node):
            if node is None:
                return None
            
            dfs(node.left)
            val_li.append(node.val)
            dfs(node.right)

        dfs(root)
        return val_li

sol = Solution()
res = sol.inorderTraversal(root)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(H)$ ($H$ is height of tree)

## Solution 2: Using stack

```python
class Solution(object):            
    def inorderTraversal(self, root):
        stack = []
        val_list = []
        current_node = root

        while stack or current_node:
            while current_node:
                stack.append(current_node)
                current_node = current_node.left

            current_node = stack.pop()
            val_list.append(current_node.val)
            current_node = current_node.right

        return val_list

sol = Solution()
res = sol.inorderTraversal(root)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(H)$ ($H$ is height of tree)
