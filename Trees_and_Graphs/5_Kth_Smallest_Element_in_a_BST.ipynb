## Preset

```python
class TreeNode(object):
    def __init__(self, val=0, left=None, right=None):
        self.val = val
        self.left = left
        self.right = right
```

## Solution 1: Using feature of BST

```python
class Solution(object):
    def kthSmallest(self, root, k):
        if not root:
            return None
        val_li = []
        def make_list(root):
            if not root:
                return None

            make_list(root.left)
            val_li.append(root.val)
            make_list(root.right)

        make_list(root)
        return val_li[k-1]
```

Time Complexity: $O(N)$

Space Complexity: $O(N)$

```python
class Solution(object):
    def kthSmallest(self, root, k):
        self.count = 0
        self.result = 0
        def smallest(root):
            if not root:
                return None
            
            smallest(root.left)
            self.count += 1
            if self.count == k:
                self.result = root.val
                return None

            smallest(root.right)
        smallest(root)
        return self.result
```

Time Complexity: $O(N)$

Space Complexity: $O(H)$ ($H$ is height of tree)
