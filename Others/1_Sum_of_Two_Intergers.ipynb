## Solution 1: Simplest Solution

```python
class Solution(object):
    def getSum(self, a, b):
        return a + b
```

Time Complexity: $O(1)$

Space Complexity: $O(1)$

## Solution 2: Using XOR and AND (Only positive number)

```python
class Solution(object):
    def getSum(self, a, b):
        while b != 0:
            carry = (a & b) << 1
            a = a ^ b
            b = carry
        return a

sol = Solution()
res = sol.getSum(10, 4)
print(res)
        
```

Time Complexity: $O(1)$

Space Complexity: $O(1)$

## Solution 2: Using MASK

```python
class Solution(object):
    def getSum(self, a, b):
        MASK = 0xFFFFFFFF
        MAX_INT = 0x7FFFFFFF
        
        while b != 0:
            carry = ((a & b) << 1) & MASK
            a = (a ^ b) & MASK
            b = carry
            
        if a > MAX_INT:
            return ~(a ^ MASK)
        else:
            return a

sol = Solution()
print(sol.getSum(-1, 1))  # 0
print(sol.getSum(-2, 3))  # 1
```

Time Complexity: $O(1)$

Space Complexity: $O(1)$
