```python
class TreeNode(object):
    def __init__(self, x, left=None, right=None):
        self.val = x
        self.left = left
        self.right = right

node4 = TreeNode(5, None, None)
node3 = TreeNode(4, None, None)
node2 = TreeNode(3, node3, node4)
node1 = TreeNode(2, None, None)
root = TreeNode(1, node1, node2)
```

## Solution 1: DFS

```python
class Codec:
    def serialize(self, root):
        if not root:
            return ''
        
        self.res = []
        def recursion(root):
            if not root:
                self.res.append('None')
                return

            self.res.append(str(root.val))
            recursion(root.left)
            recursion(root.right)

        recursion(root)
        return ','.join(self.res)

    def deserialize(self, data):
        it = iter(data.split(','))
        def recursion():
            val = next(it, None)
            if val == None or val == 'None':
                return

            node = TreeNode(int(val))
            node.left = recursion()
            node.right = recursion()

            return node
        
        return recursion()
        
codec = Codec()
res = codec.serialize(root)
print(res) 
res = codec.deserialize(res)
print(res)
```

Time Complexity of serialize: $O(N)$
<br>
Time Complexity: of deserialize: $O(N)$

Space Complexity of serialize:  $O(N)$
<br>
Space Complexity of deserialize:  $O(N)$


## Solution 2: BFS

```python
from collections import deque
class Codec:
    def serialize(self, root):
        if not root:
            return ''
        
        res = []
        queue = deque([root])

        while queue:
            curr_node = queue.popleft()
            if not curr_node:
                res.append('None')
            else:
                queue.append(curr_node.left)
                queue.append(curr_node.right)
                res.append(str(curr_node.val))

        return ','.join(res)


    def deserialize(self, data):
        if len(data) == 0:
            return None

        values = data.split(',')
        root = TreeNode(int(values[0]))
        queue = deque([root])

        i = 1
        while queue:
            curr_node = queue.popleft()

            if values[i] != 'None':
                curr_node.left = TreeNode(int(values[i]))
                queue.append(curr_node.left)
            i+=1
            if values[i] != 'None':
                curr_node.right = TreeNode(int(values[i]))
                queue.append(curr_node.right)
            i+=1

        return root

codec = Codec()
res = codec.serialize(root)
print(res) 
res = codec.deserialize(res)
print(res)
```

Time Complexity of serialize: $O(N)$
<br>
Time Complexity: of deserialize: $O(N)$

Space Complexity of serialize:  $O(N)$
<br>
Space Complexity of deserialize:  $O(N)$
