## Solution 1: Using list and dict

```python
import random

class RandomizedSet(object):

    def __init__(self):
        self.set = {}
        self.nums = []

    def insert(self, val):
        if val in self.set:
            return False
        else:
            self.set[val] = len(self.nums)
            self.nums.append(val)
            return True

        
    def remove(self, val):
        if val not in self.set:
            return False
        else:
            remove_idx = self.set[val]
            last_element = self.nums[-1]
            self.set[last_element] = remove_idx

            self.nums[remove_idx], self.nums[-1] = self.nums[-1], self.nums[remove_idx]
            self.nums.pop()
            self.set.pop(val)
            return True

    def getRandom(self):
        return random.choice(self.nums)
```

Time complexity of Each function: $O(1)$
<br>
Space complexity of Each function: $O(1)$
