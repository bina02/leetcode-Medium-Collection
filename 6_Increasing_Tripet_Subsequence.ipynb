## Solution 1: Brute force

```python
class Solution(object):
    def increasingTriplet(self, nums):
        for i in range(len(nums)-2):
            for j in range(i+1, len(nums)):
                for k in range(j+1, len(nums)):
                    if nums[i] < nums[j] < nums[k] and i < j < k:
                        return True
        return False

sol = Solution()
nums = [1,2,3,4,5]
res = sol.increasingTriplet(nums)
print(res)
```

Time Complexity: $O(N^3)$

Space Complexity: $O(1)$

## Solution 2: Using boolean matrix

```python
class Solution():
    def increasingTriplet(self, nums):
        n = len(nums)
        matrix = [[False]*n for _ in range(n)]

        # fill matrix
        for i in range(n): 
            for j in range(n):
                if nums[i] < nums[j]:
                    matrix[i][j] = True
        # convert lower triangle to False
        for i in range(n):
            for j in range(n):
                if j < i:
                    matrix[i][j] = False

        # search
        for i in range(n):
            for j in range(n):
                if matrix[i][j]:
                    if True in matrix[j]:
                        return True
        return False

sol = Solution()
nums = [20,100,10,12,5,13]
res = sol.increasingTriplet(nums)
print(res)
```

Time Complexity: $O(N^3)$

Space Complexity: $O(N^2)$

## Solution 3: Greedy algorithm

```python
class Solution():
    def increasingTriplet(self, nums):
        first = float('inf')
        second = float('inf')

        for num in nums:
            if num <= first:
                first = num
            elif num <= second:
                second = num
            else:
                return True
        return False


sol = Solution()
nums = [1,2,2147483647]
res = sol.increasingTriplet(nums)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$
