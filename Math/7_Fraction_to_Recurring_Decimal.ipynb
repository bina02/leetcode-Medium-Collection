## Solution 1: Using dict

```python
class Solution(object):
    def fractionToDecimal(self, numerator, denominator):
        if numerator == 0:
            return '0'
        result = []
        seen = {}
        
        if (numerator > 0) ^ (denominator > 0):
            result.append('-')
        numerator, denominator = abs(numerator), abs(denominator)

        result.append(str(numerator // denominator))
        remain = numerator % denominator
        if remain == 0:
            return "".join(result)

        result.append('.')

        while remain != 0:
            if remain in seen:
                result.insert(seen[remain], '(')
                result.append(')')
                break

            seen[remain] = len(result)
            remain *= 10
            result.append(str(remain // denominator))
            remain %= denominator

        return ''.join(result)

sol = Solution()
res = sol.fractionToDecimal(4,333)
print(res)
        
```

Time Complexity: $O(D)$ ($D$ is denominator)

Space Complexity: $O(D)$
