## Solution 1: Most simple solution

```python
class Solution(object):
    def myPow(self, x, n):
        return x ** n

sol = Solution()
res = sol.myPow(4, -3)
print(res)
```

Time Complexity: $O(\text{log}N)$

space Complexity: $O(1)$

## Solution 2: Brute force

```python
class Solution(object):
    def myPow(self, x, n):
        res = 1
        is_neg = False

        if n < 0:
            n = -n
            is_neg = True

        for _ in range(n):
            res *= x

        if is_neg:
            res = 1/res

        return res

sol = Solution()
res = sol.myPow(4, -3)
print(res)
```

Time Complexity: $O(N)$

space Complexity: $O(1)$

## Solution 3: Binary Exponentiation

```python
class Solution(object):
    def myPow(self, x, n):
        res = 1
        curr = x
        is_neg = False  

        if n < 0:
            n = -n
            is_neg = True

        while n > 0:
            if n % 2 == 1:
                res *= curr
            curr = curr * curr
            n //= 2

        return res if is_neg is False else 1/res

sol = Solution()
res = sol.myPow(4, 2)
print(res)
```

Time Complexity: $O(\text{log}N)$

space Complexity: $O(1)$
