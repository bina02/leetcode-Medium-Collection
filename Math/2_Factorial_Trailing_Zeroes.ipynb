## Solution 1: Get the factorial and find number of zeroes

```python
import math

class Solution(object):
    def trailingZeroes(self, n):
        count = 0
        f = math.factorial(n)
        n, digit = divmod(f, 10)
        while digit == 0 and n > 0:
            count += 1
            n, digit = divmod(n, 10)

        return count

sol = Solution()
res = sol.trailingZeroes(50)
print(res)
```

Time Complexity: $O(N\text{log}N)$

Space Complexity: $O(1)$

## Solution 2: Find number of 5

```python
class Solution(object):
    def trailingZeroes(self, n):
        count = 0
        while n >= 5:
            count += n // 5
            n //= 5

        return count


sol = Solution()
res = sol.trailingZeroes(125)
print(res)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(1)$
