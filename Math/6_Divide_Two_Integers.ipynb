## Solution 1: Simplest solution

```python
class Solution(object):
    def divide(self, dividend, divisor):
        return int(dividend / divisor)

sol = Solution()
res = sol.divide(7, -3)
print(res)
```

## Solution 2: Using subtraction

```python
class Solution(object):
    def divide(self, dividend, divisor):
        MAX_INT = 2147483647
        MIN_INT = -2147483648

        if dividend == MAX_INT and divisor == -1:
            return MIN_INT
        
        res = 0
        is_neg = (dividend > 0) ^ (divisor > 0)
        dividend, divisor = abs(dividend), abs(divisor)

        while dividend >= divisor:
            dividend -= divisor
            res += 1

        return res if is_neg is False else -res
        
sol = Solution()
res = sol.divide(1000000000, 4)
print(res)
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$

## Solution 3: Using Bitwise Shift

```python
class Solution(object):
    def divide(self, dividend, divisor):
        MAX_INT = 2147483647
        MIN_INT = -2147483648

        if dividend == MAX_INT and divisor == -1:
            return MIN_INT
        
        res = 0
        is_neg = (dividend > 0) ^ (divisor > 0)
        dividend, divisor = abs(dividend), abs(divisor)

        while dividend >= divisor:
            count = 1
            temp = divisor

            while dividend > (temp << 1):
                temp = temp << 1
                count = count << 1

            dividend -= temp
            res += count

        return res if is_neg is False else -res


sol = Solution()
res = sol.divide(1000000000, 4)
print(res)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(1)$
