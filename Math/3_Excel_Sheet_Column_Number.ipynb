## Solution 1: Positional Weighting Method

```python
class Solution(object):
    def titleToNumber(self, columnTitle):
        res = 0
        length = len(columnTitle)
        for c in columnTitle:
            res += (ord(c)-ord('A')+1) * 26**(length-1)
            length -= 1

        return res

sol = Solution()
sol.titleToNumber('BA')   
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$

## Solution 2: Horner's Method

```python
class Solution(object):
    def titleToNumber(self, columnTitle):
        res = 0
        for c in columnTitle:
            res = res*26 + (ord(c)-ord('A')+1)

        return res

sol = Solution()
sol.titleToNumber('BA')    
```

Time Complexity: $O(N)$

Space Complexity: $O(1)$
