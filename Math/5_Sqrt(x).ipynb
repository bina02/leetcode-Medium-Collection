## Solution 1: Most simple solution

```python
import math

class Solution(object):
    def mySqrt(self, x):
        return int(math.sqrt(x))

sol = Solution()
res = sol.mySqrt(4)
print(res)
```

## Solution 2: Trying all integers

```python
class Solution(object):
    def mySqrt(self, x):
        a = 0
        b = 1
        while not (a**2 <= x <= b**2):
            a += 1
            b += 1

        if b**2 == x:
            return b
        else:   
            return a
            
sol = Solution()
res = sol.mySqrt(5123)
print(res)
```

Time Complexity: $O(\sqrt{N})$

Space Complexity: $O(1)$

## Solution 3: Binary search

```python
class Solution(object):
    def mySqrt(self, x):
        if x < 2:
            return x

        left = 1
        right = x // 2
        ans = 0

        while left <= right:
            mid = (left + right) // 2
            if mid ** 2 == x:
                return mid
            elif mid ** 2 < x:
                ans = mid
                left = mid + 1
            else:
                right = mid - 1

        return ans
            
sol = Solution()
res = sol.mySqrt(25)
print(res)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(1)$
