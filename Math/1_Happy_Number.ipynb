## Solution 1: Using hashmap

```python
class Solution(object):
    def isHappy(self, n):
        def calculate(n):
            total_sum = 0
            while n > 0:
                num, digit = divmod(n, 10)
                total_sum += digit**2
                n = num
            return total_sum

        seen = set()
        n = calculate(n)

        while n != 1:
            if n not in seen:
                seen.add(n)
                n = calculate(n)
            else:
                return False
        return True
            
sol = Solution()
a = sol.isHappy(11)
print(a)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(N)$

## Solution 2: Floyd's Cycle-Finding

```python
class Solution(object):
    def isHappy(self, n):
        def calculate(n):
            total_sum = 0
            while n > 0:
                num, digit = divmod(n, 10)
                total_sum += digit**2
                n = num
            return total_sum

        slow = n
        fast = calculate(n)

        while fast != 1 and fast != slow:
            slow = calculate(slow)
            fast = calculate(calculate(fast))

        return fast == 1

sol = Solution()
a = sol.isHappy(133)
print(a)
```

Time Complexity: $O(\text{log}N)$

Space Complexity: $O(1)$
